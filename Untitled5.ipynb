```python
# ============================================================
# ANDHRA PRADESH POPULATION INTELLIGENCE LAB
# ============================================================

import numpy as np
import pandas as pd
import plotly.graph_objects as go
import plotly.express as px

from IPython.display import display, HTML, clear_output
import ipywidgets as widgets


# ============================================================
# 1. PROJECT CONFIGURATION
# ============================================================

PROJECT_TITLE = "ANDHRA PRADESH POPULATION INTELLIGENCE LAB"

AGE_GROUPS = [
    "Children (0–14)",
    "Adults (15–64)",
    "Elderly (65+)"
]

# Simplified initial population model
# Unit = million people

DEFAULT_POPULATION = np.array([
    12.0,
    31.0,
    6.4
], dtype=float)

# Simplified transition matrix
A = np.array([
    [0.20, 0.80, 0.10],
    [0.70, 0.60, 0.00],
    [0.00, 0.10, 0.90]
], dtype=float)


# ============================================================
# 2. PROFESSIONAL HEADER
# ============================================================

display(HTML("""
<div style="
background: linear-gradient(135deg,#071952,#088395,#35A29F);
padding:30px;
border-radius:18px;
color:white;
text-align:center;
font-family:Arial;
box-shadow:0 8px 25px rgba(0,0,0,0.2);
">

<h1 style="font-size:34px;margin-bottom:5px;">
ANDHRA PRADESH
</h1>

<h2 style="font-weight:400;margin-top:5px;">
Population Intelligence Lab
</h2>

<p style="font-size:17px;">
A Matrix & Eigenvalue Based Population Growth Model
</p>

<p style="font-size:14px;">
BTech First-Year Mathematics Project
</p>

</div>
"""))


# ============================================================
# 3. MATHEMATICAL MODEL
# ============================================================

display(HTML("""
<div style="
background:#f7f9fc;
padding:20px;
border-left:6px solid #088395;
border-radius:10px;
margin-top:20px;
">

<h2>🧮 Mathematical Engine</h2>

<p style="font-size:17px;">
The population is represented using three age groups:
</p>

<ul>
<li>👶 Children: 0–14 years</li>
<li>🧑 Adults: 15–64 years</li>
<li>👵 Elderly: 65+ years</li>
</ul>

<h3 style="text-align:center;">
X<sub>n+1</sub> = A X<sub>n</sub>
</h3>

<p style="text-align:center;">
The dominant eigenvalue determines long-term growth or decline,
while the dominant eigenvector represents the stable age distribution.
</p>

</div>
"""))


# ============================================================
# 4. DISPLAY TRANSITION MATRIX
# ============================================================

matrix_df = pd.DataFrame(
    A,
    index=AGE_GROUPS,
    columns=AGE_GROUPS
)

display(HTML("<h2>🔢 Population Transition Matrix</h2>"))

display(
    matrix_df.style
    .format("{:.2f}")
    .background_gradient(cmap="Blues")
    .set_caption("Transition Matrix A")
)


# ============================================================
# 5. EIGENVALUE ANALYSIS
# ============================================================

eigenvalues, eigenvectors = np.linalg.eig(A)

dominant_index = np.argmax(np.abs(eigenvalues))

dominant_eigenvalue = eigenvalues[dominant_index].real

dominant_vector = eigenvectors[:, dominant_index].real

if dominant_vector.sum() < 0:
    dominant_vector = -dominant_vector

stable_distribution = (
    dominant_vector /
    dominant_vector.sum()
) * 100


# ============================================================
# 6. STATUS
# ============================================================

if dominant_eigenvalue > 1:
    status = "GROWTH"
    status_icon = "📈"
elif dominant_eigenvalue < 1:
    status = "DECLINE"
    status_icon = "📉"
else:
    status = "STABLE"
    status_icon = "⚖️"


# ============================================================
# 7. KPI CARDS
# ============================================================

cards = f"""
<div style="
display:flex;
gap:15px;
flex-wrap:wrap;
margin:20px 0;
font-family:Arial;
">

<div style="
flex:1;
min-width:200px;
padding:22px;
background:#eef6ff;
border-radius:15px;
text-align:center;
">

<div style="font-size:30px;">λ</div>
<h3>Dominant Eigenvalue</h3>
<h2>{dominant_eigenvalue:.4f}</h2>

</div>


<div style="
flex:1;
min-width:200px;
padding:22px;
background:#eefaf3;
border-radius:15px;
text-align:center;
">

<div style="font-size:30px;">{status_icon}</div>
<h3>Model Status</h3>
<h2>{status}</h2>

</div>


<div style="
flex:1;
min-width:200px;
padding:22px;
background:#fff7e6;
border-radius:15px;
text-align:center;
">

<div style="font-size:30px;">👥</div>
<h3>Age Groups</h3>
<h2>3</h2>

</div>


<div style="
flex:1;
min-width:200px;
padding:22px;
background:#f5efff;
border-radius:15px;
text-align:center;
">

<div style="font-size:30px;">🔬</div>
<h3>Method</h3>
<h2>Matrix Model</h2>

</div>

</div>
"""

display(HTML(cards))


# ============================================================
# 8. EIGENVALUE TABLE
# ============================================================

eigen_df = pd.DataFrame({
    "Eigenvalue": eigenvalues.real,
    "Absolute Value": np.abs(eigenvalues)
})

display(HTML("<h2>🔬 Eigenvalue Spectrum</h2>"))

display(
    eigen_df.style
    .format("{:.6f}")
    .background_gradient(cmap="Purples")
)


# ============================================================
# 9. STABLE DISTRIBUTION
# ============================================================

stable_df = pd.DataFrame({
    "Age Group": AGE_GROUPS,
    "Stable Distribution (%)": stable_distribution
})

display(HTML("<h2>🎯 Long-Term Stable Age Distribution</h2>"))

display(
    stable_df.style
    .format({"Stable Distribution (%)": "{:.2f}%"})
    .background_gradient(
        subset=["Stable Distribution (%)"],
        cmap="Greens"
    )
)


# ============================================================
# 10. SIMULATION FUNCTION
# ============================================================

def simulate_population(initial_population, generations):

    X = np.array(initial_population, dtype=float)

    records = []

    for generation in range(generations + 1):

        records.append({
            "Generation": generation,
            "Children": X[0],
            "Adults": X[1],
            "Elderly": X[2],
            "Total": X.sum()
        })

        X = A @ X

    return pd.DataFrame(records)


# ============================================================
# 11. INITIAL SIMULATION
# ============================================================

GENERATIONS = 20

population_df = simulate_population(
    DEFAULT_POPULATION,
    GENERATIONS
)


# ============================================================
# 12. TOTAL POPULATION GRAPH
# ============================================================

fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=population_df["Generation"],
        y=population_df["Total"],
        mode="lines+markers",
        name="Total Population",
        line=dict(width=4)
    )
)

fig.update_layout(
    title="📈 Andhra Pradesh Population Growth Trajectory",
    xaxis_title="Generation",
    yaxis_title="Population (Million)",
    template="plotly_white",
    height=500,
    hovermode="x unified"
)

fig.show()


# ============================================================
# 13. AGE GROUP GRAPH
# ============================================================

fig = go.Figure()

for column, label in zip(
    ["Children", "Adults", "Elderly"],
    AGE_GROUPS
):

    fig.add_trace(
        go.Scatter(
            x=population_df["Generation"],
            y=population_df[column],
            mode="lines+markers",
            name=label,
            line=dict(width=3)
        )
    )

fig.update_layout(
    title="👥 Population Structure by Age Group",
    xaxis_title="Generation",
    yaxis_title="Population (Million)",
    template="plotly_white",
    height=500,
    hovermode="x unified"
)

fig.show()


# ============================================================
# 14. STABLE DISTRIBUTION CHART
# ============================================================

fig = px.pie(
    stable_df,
    names="Age Group",
    values="Stable Distribution (%)",
    title="🎯 Predicted Stable Age Distribution"
)

fig.update_traces(
    textposition="inside",
    textinfo="percent+label"
)

fig.update_layout(
    height=500
)

fig.show()


# ============================================================
# 15. POPULATION CHANGE
# ============================================================

initial_total = population_df["Total"].iloc[0]
final_total = population_df["Total"].iloc[-1]

absolute_change = final_total - initial_total

percentage_change = (
    absolute_change /
    initial_total
) * 100


# ============================================================
# 16. FINAL RESULT CARD
# ============================================================

display(HTML(f"""

<div style="
background:linear-gradient(135deg,#111827,#1f2937);
color:white;
padding:30px;
border-radius:18px;
margin-top:25px;
font-family:Arial;
">

<h2>🏆 MODEL RESULT</h2>

<div style="
display:grid;
grid-template-columns:repeat(3,1fr);
gap:20px;
text-align:center;
">

<div>
<p>Initial Population</p>
<h2>{initial_total:.2f} M</h2>
</div>

<div>
<p>Final Population</p>
<h2>{final_total:.2f} M</h2>
</div>

<div>
<p>Model Change</p>
<h2>{percentage_change:.2f}%</h2>
</div>

</div>

<hr>

<p style="font-size:16px;">
<strong>Interpretation:</strong>
The dominant eigenvalue is
<strong>{dominant_eigenvalue:.4f}</strong>,
therefore this simplified matrix model indicates
<strong>{status.lower()}</strong>.
</p>

</div>

"""))


# ============================================================
# 17. AUTOMATIC INTERPRETATION
# ============================================================

display(HTML(f"""

<div style="
background:#f8fafc;
padding:25px;
border-radius:15px;
border:1px solid #e2e8f0;
margin-top:20px;
">

<h2>💡 What Does the Mathematics Tell Us?</h2>

<p>
The model repeatedly applies the transition matrix:
</p>

<h3 style="text-align:center;">
X<sub>n+1</sub> = A X<sub>n</sub>
</h3>

<p>
The <strong>dominant eigenvalue ({dominant_eigenvalue:.4f})</strong>
controls the long-term behavior of the system.
</p>

<p>
The corresponding <strong>eigenvector</strong> provides the
stable proportion of children, adults and elderly people.
</p>

<p>
Therefore, linear algebra gives us a mathematical way to study
how an age-structured population can evolve over time.
</p>

</div>

"""))


# ============================================================
# 18. EXPORT DATA
# ============================================================

population_df.to_csv(
    "AP_population_projection.csv",
    index=False
)

stable_df.to_csv(
    "AP_stable_age_distribution.csv",
    index=False
)

eigen_df.to_csv(
    "AP_eigenvalue_analysis.csv",
    index=False
)

print("\n✅ Project datasets exported successfully!")
print("   • AP_population_projection.csv")
print("   • AP_stable_age_distribution.csv")
print("   • AP_eigenvalue_analysis.csv")


# ============================================================
# 19. PROJECT COMPLETION MESSAGE
# ============================================================

display(HTML("""

<div style="
background:linear-gradient(135deg,#0f766e,#14b8a6);
color:white;
padding:25px;
border-radius:18px;
text-align:center;
margin-top:25px;
">

<h2>🎓 MATHEMATICAL MODEL COMPLETE</h2>

<p style="font-size:17px;">
Matrices → Eigenvalues → Eigenvectors → Simulation → Visualization
</p>

<p>
<strong>Andhra Pradesh Population Intelligence Lab</strong>
</p>

</div>

"""))
```

```python
# ============================================================
# 🎛️ INTERACTIVE POPULATION SIMULATOR
# ============================================================

generation_slider = widgets.IntSlider(
    value=20,
    min=5,
    max=50,
    step=5,
    description="Generations:",
    style={"description_width": "initial"}
)

children_slider = widgets.FloatSlider(
    value=12.0,
    min=1,
    max=30,
    step=0.5,
    description="Children:",
    style={"description_width": "initial"}
)

adults_slider = widgets.FloatSlider(
    value=31.0,
    min=1,
    max=50,
    step=0.5,
    description="Adults:",
    style={"description_width": "initial"}
)

elderly_slider = widgets.FloatSlider(
    value=6.4,
    min=1,
    max=20,
    step=0.5,
    description="Elderly:",
    style={"description_width": "initial"}
)

run_button = widgets.Button(
    description="🚀 RUN SIMULATION",
    button_style="success",
    layout=widgets.Layout(width="220px", height="45px")
)

output = widgets.Output()


def run_simulation(button):

    with output:

        clear_output(wait=True)

        initial = np.array([
            children_slider.value,
            adults_slider.value,
            elderly_slider.value
        ])

        generations = generation_slider.value

        data = simulate_population(
            initial,
            generations
        )

        final = data["Total"].iloc[-1]
        initial_total = data["Total"].iloc[0]

        change = (
            (final - initial_total)
            / initial_total
        ) * 100

        print("🚀 SIMULATION COMPLETE")
        print("-" * 50)

        print(
            f"Initial population: "
            f"{initial_total:.2f} million"
        )

        print(
            f"Final population: "
            f"{final:.2f} million"
        )

        print(
            f"Change: "
            f"{change:.2f}%"
        )

        fig = go.Figure()

        fig.add_trace(
            go.Scatter(
                x=data["Generation"],
                y=data["Total"],
                mode="lines+markers",
                name="Population"
            )
        )

        fig.update_layout(
            title="Interactive Andhra Pradesh Population Simulation",
            xaxis_title="Generation",
            yaxis_title="Population (Million)",
            template="plotly_white",
            height=500
        )

        fig.show()

        display(data.round(3))


run_button.on_click(run_simulation)

display(HTML("""
<h2>🎛️ Interactive Population Laboratory</h2>
<p>
Change the starting population and number of generations,
then run the mathematical model.
</p>
"""))

display(
    widgets.VBox([
        children_slider,
        adults_slider,
        elderly_slider,
        generation_slider,
        run_button,
        output
    ])
)
```
